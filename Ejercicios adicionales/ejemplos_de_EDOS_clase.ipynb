```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import scipy.integrate as integ
from scipy.optimize import root, newton
```

```python
def f(t, y):
    return 5*y


t0, tf = 0.0, 4 # input: tiempo inicial y final 
#ts = np.linspace(t0, tf, 100)  # values of independent variable
y0 = [4]  # input:an initial condition
sol = integ.solve_ivp(f, [t0, tf], y0) #lista del tiempo inicial y final  y depues la condicion dde inicio 
print(sol)
# 'solve_ivp' returns a dictionary with information about the integration
#  sol['y'] has the 'y' values, sol['t'] the times
#  sol['t'] is the array ts, since we passed the argument `t_eval`
plt.xlabel("$t$", fontsize=16)
plt.ylabel("$y$", fontsize=16)
plt.plot(sol["t"], sol["y"][0, :],"o")
plt.show()
```

```python
def f(t, y):
    x1,x2 =y[0],y[1]
    return np.array([x1-x1*x2,-x2+x2*x1])


t0, tf = 0.00, 10.0 # input: tiempo inicial y final 
#ts = np.linspace(t0, tf, 100)  # values of independent variable
y0 = [1 , 1.5]  # input:an initial condition
sol = integ.solve_ivp(f, [t0, tf], y0) #lista del tiempo inicial y final  y depues la condicion dde inicio 
print(sol)
# 'solve_ivp' returns a dictionary with information about the integration
#  sol['y'] has the 'y' values, sol['t'] the times
#  sol['t'] is the array ts, since we passed the argument `t_eval`
plt.xlabel("$t$", fontsize=16)
plt.ylabel("$y$", fontsize=16)
plt.plot(sol["t"], sol["y"][0, :])
plt.plot(sol["t"], sol["y"][1, :]) #pintar la segunda componente de la sol
plt.show()
```

```python
def f(t, y):
    x1,x2 =y[0],y[1]
    return np.array([x2,1.5*x2-3*x1+5]) #x2=y x1=x



t0, tf = 0.00, 10.0 # input: tiempo inicial y final 
ts = np.linspace(t0, tf, 100)  # defino un array de tiempos values of independent variable
y0 = [1 , 1.5]  # input:an initial condition
sol = integ.solve_ivp(f, [t0, tf], y0, t_eval=ts) #lista del tiempo inicial y final  y depues la condicion dde inicio , añadimos t_eval del ts de puntos 
print(sol)
# 'solve_ivp' returns a dictionary with information about the integration
#  sol['y'] has the 'y' values, sol['t'] the times
#  sol['t'] is the array ts, since we passed the argument `t_eval`
plt.xlabel("$t$", fontsize=16)
plt.ylabel("$y$", fontsize=16)
plt.plot(sol["t"], sol["y"][0, :])
plt.plot(sol["t"], sol["y"][1, :]) #pintar la segunda componente de la sol
plt.show()

#represtentar las trayectorias conjuntas : cuando tengo varias componentes ç
plt.xlabel("$t$", fontsize=16)
plt.ylabel("$y$", fontsize=16)
plt.plot(sol["y"][0, :],sol["y"][1, :])
plt.show()
```

```python
def f(t, y):
    x1,x2,x3 =y[0],y[1],y[2]
    return np.array([x2,x3,t*x1+x3]) #x2=y x1=x



t0, tf = 0.00, 1 # input: tiempo inicial y final 
ts = np.linspace(t0, tf, 100)  # defino un array de tiempos values of independent variable
y0 = [1 , 2,3]  # input:an initial condition
sol = integ.solve_ivp(f, [t0, tf], y0, t_eval=ts) #lista del tiempo inicial y final  y depues la condicion dde inicio , añadimos t_eval del ts de puntos 
print(sol)
# 'solve_ivp' returns a dictionary with information about the integration
#  sol['y'] has the 'y' values, sol['t'] the times
#  sol['t'] is the array ts, since we passed the argument `t_eval`
plt.xlabel("$t$", fontsize=16)
plt.ylabel("$y$", fontsize=16)
plt.plot(sol["t"], sol["y"][0, :]) #y 
plt.plot(sol["t"], sol["y"][1, :]) # y deriv segunda pintar la segunda componente de la sol
plt.plot(sol["t"], sol["y"][2, :]) # y deriv tercera pintar la segunda componente de la sol
plt.show()

```

## Exercise 3:

The Airy function is defined as:
$$ \mathrm{Ai}(x)={\frac {1}{\pi }}\int _{0}^{\infty }\cos\left(\frac{t^3}{3} + x t \right)\,d t .$$

This function is part of what are known as **special functions**; they cannot be expressed in terms of elementary functions (trigonometric functions, quotient of polynomials, exponentials and their inverses).
These functions are implemented in the `scipy.special` module. To import them we do:
```python
from scipy.special import airy
```
The `airy` function only requires one argument: the variable `x`, which can be a `numpy` array, and returns a list with four elements, which are: in the first position, the value of the Airy function $\mathrm{Ai}$ evaluated at `x`; in the second position, the value of its derivative $\mathrm{Ai}'$ at `x`. In the third and fourth positions are, respectively, the values ​​of the Airy function $\mathrm{Bi}$ and its derivative evaluated at `x`.

For example, if we want to calculate the value $\mathrm{Ai}(2)$ we just do the following
```python
airy(2)[0]
```
We recommend that you consult the `scipy.special.airy` entry in the `scipy` manual.

## Exercise 5:

A cylindrical oil tank of radius $r$ and length $L$ is filled to depth $h$:

![](CylinderTank.png)

The resulting volume of oil in the tank is
$$V = r^2L\left(\varphi - \left(1-\frac{h}{r}\right)\sin\varphi\right) $$
where
$$\cos \varphi = 1-\frac{h}{r}.$$
We want to compute $h/r$ for a given volume $V$.

- Reformulate the problem as an equation for $\varphi$ of the form $F(\varphi)=0$. Write down explicitly the function $F$ in a *markdown* cell below.

````python
def F(phi,alpha=3/4):
    return phi-np.cos(phi)*np.sin(phi)-alpha*np.pi

xmin = 0
xmax = np.pi
xs = np.linspace(xmin, xmax, 200)
ys = F(xs)
plt.figure(figsize=(8, 8))
plt.plot(xs, ys)
plt.title("Graph of F")
plt.xlabel("phi")
plt.ylabel("F")
plt.axhline(color="k")
plt.show()
````

````python
phi_0,err = bisect(F, 1.5, 2.5,full_output=True) #si pongo full_output=True me da mas info util 
print(phi_0)
print (err)
````

## Exercise 6:

A cable is suspended as shown in the figure. 

![](Catenary.png)

Its length $s$ and the sag $h$ are related to the span $L$ by:

$$s = \frac{2}{\lambda}\sinh\left(\frac{\lambda L}{2}\right),\quad
h = \frac{1}{\lambda}\left(\cosh\left(\frac{\lambda L}{2}\right)-1\right),$$

where
- $\lambda  = w_0/T_0$,
- $w_0 =$ weight of cable per unit length,
- $T_0 =$ cable tension at $O$.

We want to compute $s$ for $L = 160\, \mathrm{m}$ and $h = 15\, \mathrm{m}$.


- Write down a numerical strategy to compute $s$ taking into account that $w_0$ and $T_0$ (or, equivalently, $\lambda$) are not known. Write your answer in a *markdown* cell below.

- Solve the resulting equation using both the bisection method and Newton's method and give the value of $s$ thus obtained. 

````python
def F(lam,L=160,h=15):
    return h*lam-np.cosh(L*lam/2)+1

xmin =-0.005
xmax = 0.005
xs = np.linspace(xmin, xmax, 200)
ys = F(xs)
plt.figure(figsize=(8, 8))
plt.plot(xs, ys)
plt.title("Graph of F")
plt.xlabel("lambda")
plt.ylabel("F")
plt.axhline(color="r")
plt.show()

lamda_0,err = bisect(F, 0.002, 0.005,full_output=True) #si pongo full_output=True me da mas info util 
print(lamda_0)
print (err)

print("s=", 2/lamda_0*np.sinh(160*lamda_0/2))
````

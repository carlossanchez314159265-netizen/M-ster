## Exercise 3:

The Airy function is defined as:
$$ \mathrm{Ai}(x)={\frac {1}{\pi }}\int _{0}^{\infty }\cos\left(\frac{t^3}{3} + x t \right)\,d t .$$

This function is part of what are known as **special functions**; they cannot be expressed in terms of elementary functions (trigonometric functions, quotient of polynomials, exponentials and their inverses).
These functions are implemented in the `scipy.special` module. To import them we do:
```python
from scipy.special import airy
```
The `airy` function only requires one argument: the variable `x`, which can be a `numpy` array, and returns a list with four elements, which are: in the first position, the value of the Airy function $\mathrm{Ai}$ evaluated at `x`; in the second position, the value of its derivative $\mathrm{Ai}'$ at `x`. In the third and fourth positions are, respectively, the values ​​of the Airy function $\mathrm{Bi}$ and its derivative evaluated at `x`.

For example, if we want to calculate the value $\mathrm{Ai}(2)$ we just do the following
```python
airy(2)[0]
```
We recommend that you consult the `scipy.special.airy` entry in the `scipy` manual.
